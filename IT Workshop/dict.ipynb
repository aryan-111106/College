```python
# create
book = {"title": "Python Basics", "author": "Guido", "year": 1991}
print(book)
```

```python
# Dictionary Items
book = {"title": "Python Basics", "author": "Guido", "year": 1991}
print(book["author"])
```

```python
# Changeable
student = {"name": "Ravi", "age": 20}
student["age"] = 21
print(student)
```

```python
# Dictionary Length
gadgets = {"phone": "iPhone", "laptop": "Dell", "tablet": "iPad"}
print(len(gadgets))
```

```python
# type()
print(type({"a": 1, "b": 2}))
```

```python
# The dict() Constructor
person = dict(name="John", city="London", age=30)
print(person)
```

```python
# Accessing Items
person = {"name": "John", "city": "London"}
print(person["city"])
print(person.get("name"))
```

```python
# Get Keys
car = {"brand": "Tesla", "model": "X", "year": 2020}
print(car.keys())
```

```python
# Get Values
car = {"brand": "Tesla", "model": "X", "year": 2020}
print(car.values())
```

```python
# Get Items
car = {"brand": "Tesla", "model": "X", "year": 2020}
print(car.items())
```

```python
# Check if Key Exists
student = {"name": "Anita", "roll": 45}
if "roll" in student:
    print("Yes, roll exists")
```

```python
# Change Values
student = {"name": "Anita", "roll": 45}
student["roll"] = 50
print(student)
```

```python
# Update Dictionary
student = {"name": "Anita", "roll": 45}
student.update({"roll": 60, "grade": "A"})
print(student)
```

```python
# Adding Items
student = {"name": "Anita", "roll": 45}
student["city"] = "Mumbai"
print(student)
```

```python
# Removing Items
student = {"name": "Anita", "roll": 45, "city": "Mumbai"}
student.pop("roll")
student.popitem()
# del student["city"]
student.clear()
print(student)
```

```python
# Loop Through a Dictionary
student = {"name": "Anita", "roll": 45, "city": "Mumbai"}
for k in student:
    print(k)
for v in student.values():
    print(v)
for k, v in student.items():
    print(k, v)
```

```python
# Copy a Dictionary
original = {"a": 1, "b": 2}
copy1 = original.copy()
copy2 = dict(original)
print(copy1, copy2)
```

```python
# Nested Dictionaries
school = {
    "class1": {"student": "Ravi", "age": 14},
    "class2": {"student": "Anita", "age": 15},
}
print(school["class1"]["student"])
```

```python
# Loop Through Nested Dictionaries
school = {
    "class1": {"student": "Ravi", "age": 14},
    "class2": {"student": "Anita", "age": 15},
}
for k, obj in school.items():
    print(k)
for subk, subv in obj.items():
    print(subk, subv)
```

```python
# setdefault()
data = {"x": 1}
print(data.setdefault("y", 5))
print(data)
```

```python
# fromkeys()
keys = ["a", "b", "c"]
new = dict.fromkeys(keys, 100)
print(new)
```
