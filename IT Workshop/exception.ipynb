```python
# except Clause

# Catching Specific Exceptions
try:
    x = int("str")  # ValueError
    inv = 1 / x
except ValueError:
    print("Not Valid!")
except ZeroDivisionError:
    print("Zero has no inverse!")

# Catching Multiple Exceptions Together
a = ["10", "twenty", 30]
try:
    total = int(a[0]) + int(a[1])
except (ValueError, TypeError) as e:
    print("Error", e)
except IndexError:
    print("Index out of range.")

# Catch-All Handler
try:
    res = "100" / 20  # TypeError
except ArithmeticError:
    print("Arithmetic problem.")
except:
    print("Something went wrong!")
```

```python
# try–finally Clause
try:
    f = open("demofile.txt")
    try:
        f.write("Lorum Ipsum")
    except:
        print("Something went wrong when writing to the file")
    finally:
        f.close()
except:
    print("Something went wrong when opening the file")
```

```python
# Raising Exceptions
def set(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    print(f"Age set to {age}")


try:
    set(-5)
except ValueError as e:
    print(e)
```

```python
# User-Defined Exceptions
class AgeError(Exception):
    pass


def set(age):
    if age < 0:
        raise AgeError("Age cannot be negative.")
    print(f"Age set to {age}")


try:
    set(-5)
except AgeError as e:
    print(e)
```

```python
# Common Built-in Exceptions with Examples
# 1. ArithmeticError
try:
    x = 10 / 0
except ArithmeticError as e:
    print("ArithmeticError:", e)
# 2. AssertionError
try:
    assert 5 > 10, "5 is not greater than 10"
except AssertionError as e:
    print("AssertionError:", e)
# 3. AttributeError
try:
    num = 10
    num.append(5)
except AttributeError as e:
    print("AttributeError:", e)
# 4. EOFError
try:
    raise EOFError
except EOFError:
    print("EOFError: No input provided.")
# 5. ImportError
try:
    import non_existing_module
except ImportError as e:
    print("ImportError:", e)
# 6. IndexError
try:
    a = [1, 2, 3]
    print(a[5])
except IndexError as e:
    print("IndexError:", e)
# 7. KeyError
try:
    d = {"a": 1}
    print(d["b"])
except KeyError as e:
    print("KeyError:", e)
# 8. MemoryError
try:
    x = [1] * (10**10)
except MemoryError:
    print("MemoryError: Too much memory used.")
# 9. NameError
try:
    print(x)
except NameError as e:
    print("NameError:", e)
# 10. NotImplementedError
try:

    class Base:
        def func(self):
            raise NotImplementedError("Must be overridden")

    Base().func()
except NotImplementedError as e:
    print("NotImplementedError:", e)
# 11. OSError
try:
    open("/path/does/not/exist.txt")
except OSError as e:
    print("OSError:", e)
# 12. OverflowError
try:
    import math

    print(math.exp(1000))
except OverflowError as e:
    print("OverflowError:", e)
# 13. ReferenceError
import weakref


class A:
    pass


a = A()
r = weakref.ref(a)
del a
print(r())  # None instead of ReferenceError
# 14. RuntimeError
try:
    raise RuntimeError("Generic runtime error")
except RuntimeError as e:
    print("RuntimeError:", e)
# 15. StopIteration
try:
    it = iter([1, 2])
    print(next(it))
    print(next(it))
    print(next(it))
except StopIteration:
    print("StopIteration: No more elements.")
# 16. SystemError
try:
    raise SystemError("Internal Python error")
except SystemError as e:
    print("SystemError:", e)
# 17. SystemExit
import sys

try:
    sys.exit()
except SystemExit:
    print("SystemExit: Program exited.")
# 18. TypeError
try:
    res = "10" + 5
except TypeError as e:
    print("TypeError:", e)
# 19. UnboundLocalError
try:

    def func():
        print(x)
        x = 10

    func()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
# 20. UnicodeDecodeError
try:
    b = b"\x80abc".decode("utf-8")
except UnicodeDecodeError as e:
    print("UnicodeDecodeError:", e)
# 21. ValueError
try:
    num = int("hello")
except ValueError as e:
    print("ValueError:", e)
# 22. ZeroDivisionError
try:
    print(5 / 0)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
```
