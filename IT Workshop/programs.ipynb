```python
#WAP to check if a string is a palindrome or not.
s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
#write a python program to convert vowel to consonant and vice versa
import random
s = input("Enter a char: ")

vowels = "aeiouAEIOU"
consonants = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"

result = ""

for ch in s:
    if ch in vowels:
        result += random.choice(consonants)   # vowel → consonant
    elif ch.isalpha():
        result += random.choice(vowels)       # consonant → vowel
    else:
        result += ch                          # spaces/symbols unchanged

print("Converted char:", result)
```

```python
#write a python program to convert vowel to consonant and vice versa in a string
import random

s = input("Enter a string: ")

vowels = "aeiouAEIOU"
consonants = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"

result = ""

for ch in s:
    if ch in vowels:
        result += random.choice(consonants)   # vowel → consonant
    elif ch.isalpha():
        result += random.choice(vowels)       # consonant → vowel
    else:
        result += ch                          # spaces/symbols unchanged

print("Converted string:", result)
```
