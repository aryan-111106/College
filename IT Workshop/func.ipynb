```python
# Arguments
def greet_user(name):
    print("Hello", name)


greet_user("Amit")
greet_user("Sara")
```

```python
# Arbitrary arguments(*args)
def list_subjects(*subjects):
    print("First subject is:", subjects[0])


list_subjects("Math", "Science", "English")
```

```python
# Keyword arguments
def display_student(name, age):
    print(f"{name} is {age} years old")


display_student(age=20, name="Ravi")
```

```python
# Arbitrary keyword arguments(**kwargs)
def student_info(**info):
    print("City:", info["city"])


student_info(name="Meena", age=21, city="Delhi")
```

```python
# Default parameter value
def country_info(country="India"):
    print("I am from", country)


country_info("Japan")
country_info()
```

```python
# Passing a list of argument
def print_numbers(nums):
    for n in nums:
        print(n)


numbers = [2, 4, 6]
print_numbers(numbers)
```

```python
# Return values
def square(x):
    return x * x


print(square(4))
print(square(9))
```

```python
# Pass statement
def future_function():
    pass
```

```python
# Positional only arguments
def area(length, width, /):
    print("Area:", length * width)


area(5, 3)
# area(length=5, width=3) ❌ error
```

```python
# Keyword only arguments
def introduce(*, name, city):
    print(f"{name} lives in {city}")


introduce(name="Arjun", city="Mumbai")
```

```python
# Combine positional and keyword only
def calculate(a, b, /, *, c, d):
    print(a + b + c + d)


calculate(2, 3, c=4, d=5)
```

```python
# Recursion
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)


print("Factorial of 5:", factorial(5))
```

```python
# Recursive Functions
def factorial(n):
    return 1 if n == 0 else n * factorial(n - 1)


factorial(5)
```

```python
# Lambda (Anonymous) Functions
square = lambda x: x * x
print(square(7))

add_five = lambda x: x + 5
print(add_five(10))

multiply = lambda a, b: a * b
print(multiply(4, 7))

total = lambda a, b, c: a + b + c
print(total(2, 5, 8))


def make_multiplier(n):
    return lambda x: x * n


mydoubler = make_multiplier(2)
mytripler = make_multiplier(3)

print(mydoubler(11))  # 22
print(mytripler(11))  # 33
```

```python
# Higher-order Functions
def apply_twice(func, val):
    return func(func(val))


print(apply_twice(lambda x: x + 1, 5))
```
